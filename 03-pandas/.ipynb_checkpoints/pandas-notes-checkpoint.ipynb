# 2. Fetching Rows and Columns from a DataFrame

```python
# fetching one column
data['winner']
```

```python
# verifying the datatype
type(data['winner'])
```

```python
# fetching multiple columns
data[['team1', 'team2', 'winner']]
```

```python
# fetching rows data.iloc[i : n-1 -> rows, i : n-1 -> cols]
data.iloc[1:3]
```

```python
# fetching multiple rows while skipping a defined range
data.iloc[1:11:2]
```

```python
# fetching desired rows through fancy indexing
data.iloc[[1,5,6]]
```

```python
# fetching rows and columns
data.iloc[:,[4,5,10]]
```

# 3. Filtering Dataframe on a Condition (Masking)

```python
mask = data['city'] == 'Hyderabad'
data[mask]
```

```python
def get_city_match_count(city):
    mask = data['city'] == city
    return data[mask].shape[0]
```

```python
get_city_match_count('Rajkot')
```

```python
mask1 = data['city'] == 'Hyderabad'
mask2 = data['date'] > '2010-01-01'
data[mask1 & mask2]
```
