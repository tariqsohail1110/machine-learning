```python
# Finding the top 5 batsmen with the most sixes scored in the past 10 years
mask = delivery['batsman_runs'] == 6
new_delivery = delivery[mask]
```

```python
new_delivery.shape[0]
```

```python
new_delivery.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head(5)
```

```python
# Finding the top 3 teams against whom V Kohli scored the most runs
def runs_scored(batsmanName):
    batsman = delivery[delivery['batsman']==batsmanName]
    return batsman.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending = False).head(1)
```

```python
runs_scored('V Kohli')
```

```python
runs_scored('MS Dhoni')
```

# 9. The isin() Function in Pandas

```python
# Scenario => Finding the most destructrive death over batsman in the history of IPL
# Strike Rate = (Number of runs / Number of balls) * 100
# Min batsman 200 balls in over 16-20
```

```python
# Extracting the death over criterea
death_over = delivery[delivery['over'] > 15]
```

```python
# Calculating the number of balls played by each batsman in death overs
all_batsman = death_over.groupby('batsman')['batsman_runs'].count()
```

```python
# Applying the more than 200 balls condition
x = all_batsman > 200
batsman_list = all_batsman[x].index.tolist() #=> boolean indexing
```

```python
# runs scored by each of the 43 batsmen
# balls played by each of the 43 batsmen
final = delivery[delivery['batsman'].isin(batsman_list)] # => isin() requires a list as parameter
```

```python
# calulating runs scored
runs = final.groupby('batsman')['batsman_runs'].sum()
```

```python
# calulating balls played
balls = final.groupby('batsman')['batsman_runs'].count()
```

```python
sr = (runs / balls) * 100
```

```python
sr.sort_values(ascending=False)
```

# 15. The dropna() method in Pandas

```python
dataframe = pd.read_csv('titanic.csv')
dataframe.shape
```

```python
dataframe.dropna() # drops rows having null values in any of their column
```

```python
dataframe.dropna(how='all') # drop the rows having missing values in all of it's columns
```

```python
dataframe.dropna(axis=1) # deletes the columns having one missing value in any one or more row(s)
```

```python
dataframe.dropna(axis=1, how='all') # deletes the columns having one missing value in all the rows
```
