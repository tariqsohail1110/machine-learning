# 4. Numpy Array vs Python Lists
## Lists vs Arrays
* Nd arrays are faster.
* They are convinient.
* The consume less memory.

## Code Examples:

```python
# lists
lista = range(100)
arr11 = np.arange(100)
```

```python
import sys
print(sys.getsizeof(87) * len(lista))
```

```python
print(arr11.itemsize * arr11.size)
```

```python
# for faster execution check
import time
x = range(10000000)
y = np.arange(10000000, 20000000)
```

```python
#lists
start_time = time.time()
c = [x + y for x,y in zip(x,y)]
print(time.time() - start_time)
```

```python
# nd arrays
a = np.arange(10000000)
b = np.arange(10000000, 20000000)
start_time = time.time()
c = a + b
print(time.time() - start_time)
```

# 11. Broadcasting in Numpy
To perform operations on different shaped arrays, we use broadcasting in numpy

## Code Examples:

```python
# If x=m  and y=n, operation will take place
a1 = np.arange(8).reshape(2,4)
a2 = np.arange(8,16).reshape(2,4)

a1 + a2
```

```python
# If x=1 and y=n, then also operation will take place (Same Dimensions)
a3 = np.arange(3).reshape(1,3)
a4 = np.arange(12).reshape(4,3)

a3 + a4
```

```python
# If y=1 and x=m, operation will take place even not of same dimensions
a5 = np.arange(4).reshape(4,1)
a6 = np.arange(12).reshape(4,3)

a5 + a6
```

```python
# If x=1 and y!=n, then operation will not take place
a7 = np.arange(3).reshape(1,3)
a8 = np.arange(16).reshape(4,4)

a7 + a8
```

```python
# If x=1 and y=1, then y==m, operation will take place
a9 = np.arange(3).reshape(1,3)
a10 = np.arange(3).reshape(3,1)

a9 + a10
```

```python
# If x=1 and y=1, no matter what, operation will occur
a11 = np.arange(1).reshape(1,1)
a12 = np.arange(20).reshape(4,5)

a11 + a12
```

```python
# Even if they're of different dimensions
a13 = np.arange(4)
a14 = np.arange(20).reshape(5,4)

a13 + a14
```
