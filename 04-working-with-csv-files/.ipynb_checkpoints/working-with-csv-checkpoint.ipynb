# 3. Opening a csv file from a URL

```python
import requests
from io import StringIO

# any url to the csv file will be inserted below
url = "https://raw.githubusercontent.com/google/dspl/master/samples/google/canonical/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macimtosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

df = pd.read_csv(data)
df.head()
```

# 9. Skiprows / nrows Parameter

# 10. Encoding Parameter

```python
df = pd.read_csv('Fortune500.csv', encoding='utf-8') #define the encoding of the actual file
df.head()
```

# 11. Skip Bad Lines

```python
df = pd.read_csv('Fortune500.csv', sep=',', encoding='utf-8', on_bad_lines='skip') # on_bad_lines is used to skip bad lines while loading
df
```

# 12. Dtypes Parameter

```python
# df = pd.read_csv('Fortune500.csv', dtype={'profits':in})
df = pd.read_csv('Fortune500.csv')
df['profits'] = pd.to_numeric(df['profits'], errors='coerce').round().astype('Int64') 
df.head()
```

# 13. Handling Dates

```python
df = pd.read_csv('matches.csv', parse_dates=['date']).info()
df
```

# 14. Convertors

```python
def rename(name):
    abb = [word[0] for word in name.split()]
    return ("".join(abb))
```

```python
pd.read_csv('matches.csv', converters={'team1' : rename, 'team2' : rename})
```

# 15. na_values Parameters
